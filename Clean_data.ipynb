```python
import pandas as pd 
data=pd.read_excel('Churn_Unclean_Project.xlsx')
data.head(10)
```

```python
data.drop_duplicates(inplace=True)
```

```python
data.duplicated().sum()
```

```python
import numpy as np
```

```python
data.replace(["NA", "Null", "", " "],np.nan,inplace=True)
data.head(10)
```

```python
# Remove Extra Spaces
# Standardize Churn Column
# Convert Numeric Columns
# Remove Invalid Age
# Remove Negative Charges
# Convert Date
# Fill Missing Values
# Feature Engineering Customer Lifetime Value
# Tenure Group, Monthly Revenue, Senior Citizen Flag, Churn Flag
```

```python
data.apply(lambda x:x.str.strip() if x.dtype=='object' else x)
```

```python
columns=["Age", "Tenure_Months", "Monthly_Charges", "Total_Charges"]

for c in columns:
    data[c]=pd.to_numeric(data[c],errors="coerce")
```

```python
data
```

```python
data=data[(data["Age"]>=18)&(data["Age"]<=100)]
```

```python
data=data[data["Monthly_Charges"]>=0]
data=data[data["Total_Charges"]>=0]
```

```python
data.head()
```

```python

data['Gender'] = data['Gender'].str.strip().str.capitalize()
data['State'] = data['State'].str.strip().str.lower().str.capitalize()
data['Subscription_Type'] = data['Subscription_Type'].str.strip().str.lower().str.capitalize()
```

```python
data["Tech_Support"]=data["Tech_Support"].fillna("No")
data["Payment_Method"]=data["Payment_Method"].fillna("Unknown")
data["Subscription_Type"]=data["Subscription_Type"].fillna("No")
data["Monthly_Charges"]=data["Monthly_Charges"].fillna(data["Monthly_Charges"].mean())
data["Total_Charges"]=data["Total_Charges"].fillna(data["Total_Charges"].mean())
data["Age"]=data["Age"].fillna(data["Age"].mean())
```

```python
data["Customer_Value"]=data["Monthly_Charges"]*data["Total_Charges"]
```

```python
data.head()
```

```python
data["Tenure_Group"]=pd.cut(data["Tenure_Months"],
                            bins=[0,12,24,48,72],
                            labels=["0-12","13-24","25-48","49-72"]
                           )

```

```python
data["Senior_Citizen"]=np.where(data["Age"]>=60,"Senior","Adult")
```

```python
data["Churn_Flag"]=data["Churn"].map({"Yes":1,"No":0})
```
